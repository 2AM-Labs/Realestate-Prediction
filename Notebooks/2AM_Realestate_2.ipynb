```python
import folium
import pandas as pd


# Initialize the map
LAT = df['lat']
LONG = df['long']
villages = df['Village'].unique()
peta = folium.Map(location=[df['lat'].mean(), df['long'].mean()], zoom_start=15)

unique_villages = df['Village'].unique()
colors = ['red', 'green', 'yellow', 'blue']

village_color_dict = {village: colors[i % len(colors)] for i, village in enumerate(unique_villages)}


for i in range(len(LAT)):
    village = df['Village'].iloc[i]
    color = village_color_dict[village]

    folium.CircleMarker(
        location=[LAT.iloc[i], LONG.iloc[i]],
        radius=5,  # Marker size
        color=color,  # Marker color
        fill=True,
        fill_color=color,
        fill_opacity=0.6,
        popup=f'Harga: {df["price"].iloc[i]:.2f}, \nDaerah: {village}'  # Information popup
    ).add_to(peta)

# Show the map
peta
```

# **Test Linear Regression df**

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
import pandas as pd

X1 = df3.drop(['price','No'],axis='columns')
y1 = df3.price

X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.2, random_state=0)

# Buat model regresi linear
reg = LinearRegression()
reg.fit(X1_train, y1_train)

# Prediksi harga
y1_pred = reg.predict(X1_test)

# Evaluasi model
print('Linear Regression')
print('RMSE:', mean_squared_error(y1_test, y1_pred))
print('MAE:', mean_absolute_error(y1_test, y1_pred))
print('R^2:', r2_score(y1_test, y1_pred))
```

```python
import pandas as pd
import numpy as np
import pickle

# Load model regresi linear yang sudah dilatih
with open('/content/drive/MyDrive/Model/realestate_LR.pickle', 'rb') as file:
    loaded_model = pickle.load(file)

# Fungsi untuk memprediksi harga berdasarkan fitur yang diberikan
def predict_price_lr(date, age, dist, store, Dapinglin, Fulfilling, New_store, Pond):
    # Persiapkan data input dengan format yang sesuai dengan model regresi linear
    input_data = np.array([[date, age, dist, store, Dapinglin, Fulfilling, New_store, Pond]])

    # Lakukan prediksi harga menggunakan model regresi linear yang sudah dilatih
    predicted_price = loaded_model.predict(input_data)

    return predicted_price[0]
```
