```python
import networkx as nx
import matplotlib.pyplot as plt

# Buat objek graf untuk merepresentasikan CLD
G = nx.DiGraph()

# Tambahkan variabel-variabel atau faktor-faktor sebagai simpul (node) dalam CLD
G.add_node("Pengetahuan dan Pengalaman Agen Properti (K1)")
G.add_node("Data Transaksi (K2)")
G.add_node("Pemasaran Properti (F1)")
G.add_node("Penyewa/Pembeli (F2)")
G.add_node("Negosiasi (F3)")
G.add_node("Prediksi Harga (F4)")
G.add_node("Kesepakatan Harga (K3)")

# Tambahkan hubungan antara variabel-variabel dengan menggunakan tepi (edge)
# Contoh hubungan positif (+) dan negatif (-)
G.add_edge("Pengetahuan dan Pengalaman Agen Properti (K1)", "Prediksi Harga (F4)", weight=1)
G.add_edge("Prediksi Harga (F4)", "Penyewa/Pembeli (F2)", weight=1)
G.add_edge("Pemasaran Properti (F1)", "Penyewa/Pembeli (F2)", weight=1)
G.add_edge("Penyewa/Pembeli (F2)", "Prediksi Harga (F4)", weight=-1)
G.add_edge("Negosiasi (F3)", "Kesepakatan Harga (K3)", weight=1)
G.add_edge("Kesepakatan Harga (K3)", "Data Transaksi (K2)", weight=1)
G.add_edge("Data Transaksi (K2)", "Prediksi Harga (F4)", weight=-1)

# Buat posisi simpul untuk tampilan yang lebih baik
pos = nx.spring_layout(G)

# Tampilkan CLD
plt.figure(figsize=(10, 8))
nx.draw(G, pos, with_labels=True, node_size=1000, node_color="lightblue", font_size=10, font_color="black")
labels = nx.get_edge_attributes(G, "weight")
edge_labels = {(k[0], k[1]): v for k, v in labels.items()}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=10, font_color="red")
plt.title("Causal Loop Diagram")
plt.show()
```

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# Baca dataset dari file CSV
data = pd.read_csv('Realestate.csv')

# Membuat graf untuk CLD
G = nx.DiGraph()

# Tambahkan simpul (node) yang mewakili variabel dalam CLD
for column in data.columns:
    G.add_node(column)

# Tambahkan hubungan antar variabel berdasarkan data Anda
# Ini hanya contoh, Anda perlu menyesuaikan sesuai dengan dataset Anda
G.add_edge("Variabel A", "Variabel B", weight=1)
G.add_edge("Variabel B", "Variabel C", weight=-1)
G.add_edge("Variabel C", "Variabel A", weight=1)

# Buat posisi simpul untuk tampilan yang lebih baik
pos = nx.spring_layout(G)

# Tampilkan CLD
plt.figure(figsize=(10, 8))
nx.draw(G, pos, with_labels=True, node_size=1000, node_color="lightblue", font_size=10, font_color="black")
labels = nx.get_edge_attributes(G, "weight")
edge_labels = {(k[0], k[1]): v for k, v in labels.items()}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=10, font_color="red")
plt.title("Causal Loop Diagram")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Buat objek graf untuk merepresentasikan CLD
G = nx.DiGraph()

# Tambahkan variabel-variabel sebagai simpul (node) dalam CLD
variables = [
    "Harga Rumah",
    "Demand",
    "Inflasi",
    "Umur Bangunan",
    "Jarak ke Transportasi Umum",
    "Banyaknya Toko Konvensional",
    "Luas Bangunan"
]

for variable in variables:
    G.add_node(variable)

# Tambahkan hubungan antara variabel-variabel dengan menggunakan tepi (edge)
# Tentukan apakah hubungan tersebut reinforcing (+) atau balancing (-)
# Ini adalah contoh hubungan, Anda dapat menyesuaikan sesuai kebutuhan
G.add_edge("Demand", "Harga Rumah", weight=+1)  # Peningkatan Demand -> Peningkatan Harga Rumah (+)
G.add_edge("Inflasi", "Harga Rumah", weight=+1)  # Peningkatan Inflasi -> Peningkatan Harga Rumah (+)
G.add_edge("Umur Bangunan", "Harga Rumah", weight=-1)  # Peningkatan Umur Bangunan -> Penurunan Harga Rumah (-)
G.add_edge("Jarak ke Transportasi Umum", "Harga Rumah", weight=-1)  # Peningkatan Jarak -> Penurunan Harga Rumah (-)
G.add_edge("Banyaknya Toko Konvensional", "Harga Rumah", weight=+1)  # Peningkatan Toko -> Peningkatan Harga Rumah (+)
G.add_edge("Luas Bangunan", "Harga Rumah", weight=+1)  # Peningkatan Luas Bangunan -> Peningkatan Harga Rumah (+)

# Buat posisi simpul untuk tampilan yang lebih baik
pos = nx.spring_layout(G)

# Tampilkan CLD
plt.figure(figsize=(10, 8))
nx.draw(G, pos, with_labels=True, node_size=1000, node_color="lightblue", font_size=10, font_color="black")
labels = nx.get_edge_attributes(G, "weight")
edge_labels = {(k[0], k[1]): v for k, v in labels.items()}
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=10, font_color="red")
plt.title("Causal Loop Diagram")
plt.show()
```
