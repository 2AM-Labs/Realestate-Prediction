```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Realestate.csv')
```

```python
df.describe()
```

```python
df.head()
```

```python
df.info()
```

```python
corr = df.drop(['No'], axis=1).corr()
plt.subplots(figsize = (20, 20))
sns.heatmap(corr, annot = True)
```

```python
corr = df.drop(['No','Y house price of unit area'], axis=1)
corr.head()
```

```python
y = df['Y house price of unit area']
y.head( )
```

```python
from sklearn.model_selection import train_test_split
```

```python
corr_train, corr_test, y_train, y_test = train_test_split(corr,y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
lr = LinearRegression()
```

```python
lr.fit(corr_train,y_train)
```

```python
y_pred = lr.predict(corr_test)
```

```python
print('Linear Regression')
print('RMSE :', mean_squared_error(y_test, y_pred))
print('MAE :', mean_absolute_error(y_test, y_pred))
print('R :', r2_score(y_test,y_pred)*100)
```
